## Average Delivery Time - df_orders_delivered + order_purchase_timestamp + order_delivered_customer_date

```python
# delivery_days = delivered date - purchase date, in fractional days
# 8 delivered orders have no order_delivered_customer_date -> NaN, left as is (excluded from mean/median)
# Distribution is right-skewed (mean 12.6 vs median 10.2; p99 = 46 days, max = 209 days)
# Outliers kept; median reported alongside the mean
df_orders_delivered["delivery_days"] = (df_orders_delivered["order_delivered_customer_date"] - df_orders_delivered["order_purchase_timestamp"]).dt.total_seconds() / 86400
```

```python
df_orders_delivered.describe()
```

```python
print(df_orders_delivered["delivery_days"].quantile([0.9, 0.95, 0.99]))
df_orders_delivered["delivery_days"].describe()
```

## Delivery Time vs Review Score - df_orders_delivered + df_reviews_unique

```python
# 646 delivered orders have no review; dropped by inner merge
df_delivery_reviews = df_orders_delivered.merge(df_reviews_unique, on="order_id", how="inner")
```

```python
print(df_delivery_reviews.shape[0])
df_delivery_reviews["order_id"].is_unique
```

```python
df_delivery_reviews["delivery_days"].isna().sum()
```

```python
df_delivery_reviews_dropped_nan = df_delivery_reviews.dropna(subset=["delivery_days"])
df_delivery_reviews_dropped_nan["delivery_days"].isna().sum()
```

```python
# Delivered orders with a review, one review per order (latest kept).
# Median delivery time rises as score falls: 5-star 9.2 days -> 1-star 16.8 days.
# Mean-median gap also widens (1.4 -> 4.5 days): long delays concentrate in low scores.
# Association only; this does not show that delay causes the low score.
df_delivery_reviews.groupby("review_score")["delivery_days"].aggregate(["median", "mean", "count"])
```

## Late Delivery vs Review Score - df_delivery_reviews_dropped_nan

```python
df_delivery_reviews_dropped_nan["order_estimated_delivery_date"].dt.normalize()
```

```python
# is_late = delivered date (normalized to day) > estimated delivery date
# 6.7% of delivered, reviewed orders arrived late (6381 of 95824)
# Mean review score: on time 4.29 (median 5), late 2.27 (median 1)
# Association only; does not show that lateness causes low scores
df_delivery_reviews_dropped_nan["is_late"] = (df_delivery_reviews_dropped_nan["order_delivered_customer_date"].dt.normalize() > df_delivery_reviews_dropped_nan["order_estimated_delivery_date"]
)

print(df_delivery_reviews_dropped_nan["is_late"].mean())
df_delivery_reviews_dropped_nan.groupby("is_late")["review_score"].aggregate(["median", "mean", "count"])
```
